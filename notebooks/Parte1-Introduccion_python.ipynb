# Listas
Estructuras que almacenan diversos tipos de datos usando **corchetes**. El índice comienza en 0.

```python
mi_lista = ["image", "python", "numpy", "pandas"]
#print(mi_lista)
#mi_lista[0]
mi_lista
```

```python
mi_lista.sort() # Ordena la lista alfabéticamente
print(mi_lista) # Imprime la lista ya ordenada
```

```python
mi_lista.append("matplotlib")  #agregar un elemento a la lista
print(mi_lista)
# este código ordena los elementos de la lista de A a Z
mi_lista.sort()
# print(mi_lista)
```

```python
mi_lista[1:3] # extrae los elementos de la posición 1 a 2 - NO INCLUYE EL ÚLTIMO (SLICING)
mi_lista[1:] # extrae los elementos de la posición 1 hasta el final
```

```python
# imprimir todos los elementos de una lista
# Identación
#for ________ in _______:
for elemento in mi_lista:
  print(elemento)
  #print("Hola")
```

## Control de flujo y ciclos (bucles)
Python utiliza la indentación (4 espacios) para definir bloques de código,

```python
# Sentencias if-else (condicionales)
if 1 < 2:
    print("Primero")
elif 3 == 3:
    print("Mitad")
elif 4 == 4:
    print("Medio")
else:
    print("Ultimo")

# Ciclos For
notas = [80, 75, 65, 55, 90]
for item in notas:
    print(item)
```

# Funciones
Son bloques de código que reciben parámetros, realizan algunas tareas y retornan valores.

Se definen con `def` y usan `return` para devolver valores

```python
def sumar_todos_los_elementos(lista: list) -> float:
    #Suma todos los elementos de una lista
    suma = 0
    for valor in lista:
        suma = suma + valor
    return suma

mi_otra_lista1 = [10, 20, 30, 50]
resultado1 = sumar_todos_los_elementos(mi_otra_lista1)
mi_otra_lista2 = [5, 41, 50, 50]
resultado2 = sumar_todos_los_elementos(mi_otra_lista2)
print(resultado1)
print(resultado2)
```
